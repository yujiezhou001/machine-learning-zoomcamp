```python
X = df[df.origin == 'Asia'][['vehicle_weight', 'model_year']].head(7).values
X
```

```python
XTX = np.dot(X.T, X)
XTX
```

```python
XTX_inverse = np.linalg.inv(XTX)
XTX_inverse
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = np.dot(np.dot(XTX_inverse, X.T), y)
w
```
